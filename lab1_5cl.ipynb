```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
```

```python
d0_array = np.array([23, 46, 55, 21, 40.5, 21.5, 52.5, 24, 29.5, 37, 58.5])
di_array = np.array([52, 6.5, 24, 2.2, 27, 3, 23.5, 52, 36, 28.5, 22.5])
hi_array = np.array([4.9, 3.3, 1.2, 0.8, 1.5, 1.3, 1, 5, 2.8, 1.8, 0.9])
h0 = 2.3
```

```python
x = 1/d0_array
y = 1/di_array
```

```python
#determine covariance and coefficient of linear correlation
cov_xy = np.cov(x,y)[0]
rho_xy = np.corrcoef(x,y)[0][1]
print("sigma_xy: \n",cov_xy)
print("rho_xy: \n",rho_xy)
```

```python
alpha_x = 1.5 / (d0_array**2) #error of the meterstick (0.5 cm) + 1cm human error
alpha_y = 1.1 / (di_array**2) #error of the ruler (0.1 cm) + 1cm human error
alpha_total = np.sqrt(alpha_x**2 + alpha_y**2)
print(alpha_total)
```

```python
weights = 1/(alpha_total**2)
b = np.sum(weights * (y + x)) / np.sum(weights)
alpha_b = 1 / np.sqrt(np.sum(weights))
print(b)
print(alpha_b)
```

```python
y_model = -1 * x + b
chi2 = np.sum(((y - y_model) / alpha_total)**2)
dof = len(y) - 1
chi2_reduced = chi2 / dof
print(chi2)
print(dof)
print(chi2_reduced)
```

```python
focal = 1 / b
alpha_f = alpha_b / (b**2)
print(focal)
print(alpha_f)
```
